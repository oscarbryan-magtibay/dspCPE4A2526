```python
from openai import OpenAI # 1. Import the new 'OpenAI' client

# 2. Initialize the client with your LM Studio settings
# The base_url and api_key are now set here
client = OpenAI(
    base_url="http://127.0.0.1:1234/v1",
    api_key="lm-studio", # or "not-needed"
)

def ask_ai(prompt):
    # 3. Use the new 'client.chat.completions.create' method
    response = client.chat.completions.create(
        model="tinyllama-1.1b-chat-v1.0",  # Make sure this model is loaded in LM Studio
        messages=[
            {"role": "user", "content": prompt}
        ]
    )
    # 4. Access the content directly using .content
    return response.choices[0].message.content

print(ask_ai("How many finger in a normal human hand?"))
```
