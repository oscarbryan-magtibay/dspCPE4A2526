```python
import cv2
import pytesseract
from matplotlib import pyplot as plt

# --- OPTIONAL: If tesseract is not in PATH, set the path manually ---
# For Windows, uncomment and adjust:
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

def ocr_image(image_path):
    # Load the image
    img = cv2.imread(image_path)

    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Apply thresholding
    _, thresh = cv2.threshold(gray, 150, 5, cv2.THRESH_BINARY)

    # Morphological noise removal
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (1, 1))
    cleaned = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)

    # OCR with Tesseract
    text = pytesseract.image_to_string(cleaned)

    # Show original and processed image inline (Jupyter)
    fig, axs = plt.subplots(1, 2, figsize=(12, 6))
    axs[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    axs[0].set_title("Original Image")
    axs[0].axis("off")

    axs[1].imshow(cleaned, cmap="gray")
    axs[1].set_title("Processed for OCR")
    axs[1].axis("off")

    plt.show()

    return text


# Example usage
text_output = ocr_image("5.jpg")
print("Extracted Text:\n", text_output)
```

```python
import cv2
import pytesseract
import re
from matplotlib import pyplot as plt

# Input image
image_path = "5.jpg"
results = model(image_path, conf=0.5)  
boxes = results[0].boxes
img = cv2.imread(image_path)
img_draw = img.copy()

plate_texts = []

# Draw YOLO detections
for box in boxes:
    x1, y1, x2, y2 = map(int, box.xyxy[0])
    cv2.rectangle(img_draw, (x1, y1), (x2, y2), (0, 255, 0), 2)

plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(img_draw, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("YOLO Detection: plateno.5")
plt.show()

# Process each detected plate
for idx, box in enumerate(boxes, start=1):
    x1, y1, x2, y2 = map(int, box.xyxy[0])

    # Add padding around the box
    pad = 5
    x1, y1 = max(0, x1 - pad), max(0, y1 - pad)
    x2, y2 = min(img.shape[1], x2 + pad), min(img.shape[0], y2 + pad)

    plate = img[y1:y2, x1:x2]
    if plate.size == 0:
        print(f"Plate {idx}: empty crop (skipped)")
        plate_texts.append("")
        continue

    # Preprocessing
    gray = cv2.cvtColor(plate, cv2.COLOR_BGR2GRAY)
    gray = cv2.bilateralFilter(gray, 11, 17, 17)  # keep edges sharp

    # Adaptive thresholding (better than Otsu for uneven lighting)
    thresh = cv2.adaptiveThreshold(
        gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY, 19, 9
    )

    # Optional: thicken characters
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
    cleaned = cv2.dilate(thresh, kernel, iterations=1)

    # OCR
    config = '--psm 7 -c tessedit_char_whitelist=ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789'
    raw_text = pytesseract.image_to_string(cleaned, config=config).strip()

    # Clean result
    clean_text = re.sub(r'[^A-Z0-9]', '', raw_text.upper())

    # Validation (basic Philippine plate formats)
    if re.match(r'^[A-Z]{3}[0-9]{4}$', clean_text):  # ABC1234
        final_plate = clean_text
    elif re.match(r'^[A-Z]{4}[0-9]{4}$', clean_text):  # XABC1234 → drop first letter
        candidate = clean_text[1:]
        final_plate = candidate if re.match(r'^[A-Z]{3}[0-9]{4}$', candidate) else clean_text
    else:
        final_plate = clean_text

    plate_texts.append(final_plate)

    # Show cropped vs processed
    fig, axs = plt.subplots(1, 2, figsize=(10, 4))
    axs[0].imshow(cv2.cvtColor(plate, cv2.COLOR_BGR2RGB))
    axs[0].set_title("Cropped")
    axs[0].axis("off")

    axs[1].imshow(cleaned, cmap="gray")
    axs[1].set_title("Processed for OCR")
    axs[1].axis("off")
    plt.show()

    # Save debug images
    cv2.imwrite(f"plate_{idx}_raw.jpg", plate)
    cv2.imwrite(f"plate_{idx}_processed.jpg", cleaned)

# Final results
print("Detected Plate Numbers:")
for i, t in enumerate(plate_texts, 1):
    print(f"{i}. {t if t else '[No text detected]'}")
```
