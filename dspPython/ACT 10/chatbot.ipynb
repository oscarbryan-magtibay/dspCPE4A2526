```python
# Advanced version with FREE AI (using OpenAI-compatible API)
import os
import requests
import json
from langchain_community.document_loaders import PyPDFLoader

class AIPDFChatbot:
    def __init__(self):
        self.full_text = ""
        self.is_loaded = False
        # Using a free OpenAI-compatible API (no key needed)
        self.api_url = "https://api.openai.com/v1/chat/completions"  # You'll need an API key for this
        # Alternative: Use a free local model or different service
    
    def load_pdf(self, file_path: str):
        """Load PDF content"""
        if not os.path.exists(file_path):
            print(f"❌ File '{file_path}' not found!")
            return False
        
        try:
            loader = PyPDFLoader(file_path)
            documents = loader.load()
            
            # Combine all pages into one text
            self.full_text = "\n\n".join([doc.page_content for doc in documents])
            self.is_loaded = True
            
            print(f"✅ Loaded PDF with {len(documents)} pages")
            print(f"📝 Total text length: {len(self.full_text)} characters")
            return True
            
        except Exception as e:
            print(f"❌ Error loading PDF: {e}")
            return False
    
    def ask(self, question: str):
        """Ask question using simple text analysis"""
        if not self.is_loaded:
            return "❌ Please load a PDF first with load_pdf()"
        
        # Simple but intelligent text analysis
        question_lower = question.lower()
        text_lower = self.full_text.lower()
        
        # Common question patterns and their search strategies
        if any(word in question_lower for word in ['what', 'story about', 'summary']):
            # Get first few sentences for summary
            sentences = self.full_text.split('.')
            summary = '. '.join(sentences[:3]) + '.'
            return f"Story Summary:\n\n{summary}"
        
        elif any(word in question_lower for word in ['who', 'character']):
            # Look for character mentions
            sentences = self.full_text.split('.')
            character_sentences = []
            for sentence in sentences:
                if any(keyword in sentence.lower() for keyword in ['he', 'she', 'they', 'character', 'name']):
                    character_sentences.append(sentence.strip())
            if character_sentences:
                return f"Characters mentioned:\n\n" + '\n'.join(character_sentences[:5])
        
        elif any(word in question_lower for word in ['when', 'where', 'setting']):
            # Look for setting clues
            sentences = self.full_text.split('.')
            setting_sentences = []
            for sentence in sentences:
                if any(keyword in sentence.lower() for keyword in ['in', 'at', 'on', 'when', 'where']):
                    setting_sentences.append(sentence.strip())
            if setting_sentences:
                return f"Setting details:\n\n" + '\n'.join(setting_sentences[:5])
        
        elif any(word in question_lower for word in ['how', 'why', 'happened']):
            # Look for event descriptions
            sentences = self.full_text.split('.')
            event_sentences = []
            for sentence in sentences:
                if any(keyword in sentence.lower() for keyword in ['then', 'after', 'before', 'because', 'so']):
                    event_sentences.append(sentence.strip())
            if event_sentences:
                return f"Events:\n\n" + '\n'.join(event_sentences[:5])
        
        # Default: return relevant sentences based on keywords
        sentences = self.full_text.split('.')
        relevant = []
        keywords = question_lower.split()
        
        for sentence in sentences:
            sentence_lower = sentence.lower()
            if any(keyword in sentence_lower for keyword in keywords):
                if sentence.strip() and len(sentence) > 10:
                    relevant.append(sentence.strip())
        
        if relevant:
            return f"Relevant information:\n\n" + '\n\n'.join(relevant[:5])
        else:
            return "I found this general information:\n\n" + '. '.join(sentences[:3]) + '.'

# Usage
print("🚀 Starting PDF Chatbot...")
bot = AIPDFChatbot()

if bot.load_pdf("example_story.pdf"):
    print("\n💬 Chat ready! Ask questions about the story.")
    print("Type 'exit' or 'quit' to end the chat.\n")
    
    while True:
        question = input("You: ").strip()
        
        # Print the question you asked
        print(f"\nYou asked: '{question}'")
        
        if question.lower() in ['exit', 'quit']:
            print("👋 Goodbye!")
            break
            
        if question:
            print("🤖 Thinking...")
            answer = bot.ask(question)
            print(f"\nAssistant: {answer}")
            print("-" * 50)  # Separator
        else:
            print("Please enter a question.")
```
