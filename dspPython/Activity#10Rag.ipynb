```python
import google.generativeai as genai
import fitz  # PyMuPDF for reading PDFs
import os

genai.configure(api_key="AIzaSyCcMQGvlqpeZ6NtZD9M3hPf__RhNDBdwM8")

def read_pdf_file(file_path):
    """Reads text content from a PDF file using PyMuPDF."""
    try:
        if not os.path.exists(file_path):
            print(f" File not found: {file_path}")
            return ""

        text = ""
        with fitz.open(file_path) as pdf:
            for page in pdf:
                text += page.get_text("text")

        if not text.strip():
            print(" The PDF file has no readable text (may be scanned or image-only).")
        else:
            print(f" Successfully extracted text from '{os.path.basename(file_path)}'.\n")

        return text.strip()

    except Exception as e:
        print(f" Error reading PDF file: {e}")
        return ""

def ask_question_about_pdf(text):
    """Start terminal-based Q&A about the PDF."""
    print("\n You can now ask questions about the PDF.")
    print("Type 'exit' to quit.\n")

    model = genai.GenerativeModel("gemini-2.0-flash")
    chat = model.start_chat(history=[])

    while True:
        question = input(" Ask: ")
        if question.lower() in ["exit", "quit"]:
            print(" Goodbye!")
            break

        prompt = f"""
You are a helpful assistant. Based on the story below, answer the question accurately.

--- PDF CONTENT START ---
{text[:20000]}  # Limit to prevent overload
--- PDF CONTENT END ---

Question: {question}
"""

        try:
            response = chat.send_message(prompt)
            print("\n Answer:", response.text, "\n")
        except Exception as e:
            print(f" Error communicating with Gemini API: {e}")

if __name__ == "__main__":
    pdf_path = r"docs/Shrek_Summary.pdf"

    print(" Reading PDF file...\n")
    text = read_pdf_file(pdf_path)

    if text:
        print(" Text preview (first 500 chars):\n")
        print(text[:500])
        ask_question_about_pdf(text)
    else:
        print(" No readable text found in the PDF.")
```
