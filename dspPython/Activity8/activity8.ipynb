```python
import os
from ultralytics import YOLO

# Absolute path to data.yaml
data_path = os.path.join(os.getcwd(), "Musical-Instruments-Detection-1", "data.yaml")
print("Using dataset config:", data_path)

# Verify file exists
if not os.path.exists(data_path):
    raise FileNotFoundError(f"data.yaml not found at: {data_path}")

# Load YOLOv8 model
model = YOLO("yolov8n.pt")

# Train
results = model.train(
    data=data_path,
    epochs=25,
    imgsz=640,
    batch=16,
    plots=True
)

print("✅ Training finished! Results saved in:", results.save_dir)
```

```python
# CELL 5: Object Detection using your trained YOLOv8 model

from ultralytics import YOLO
import cv2
import os
import matplotlib.pyplot as plt

# --- Step 1: Find your trained model (best.pt) ---
best_model_path = None
for root, dirs, files in os.walk("runs/detect"):
    for f in files:
        if f == "best.pt":
            best_model_path = os.path.join(root, f)

if best_model_path is None:
    raise FileNotFoundError("❌ No trained model (best.pt) found in runs/detect/ directory.")
else:
    print("✅ Using trained model:", best_model_path)

# --- Step 2: Define image path ---
image_path = r"c:\Users\Thinkpad\Documents\GitHub\dspCPE4A2526\dspPython\Activity8\sample_image.jpg"
if not os.path.exists(image_path):
    raise FileNotFoundError(f"❌ Sample image not found at: {image_path}")

# --- Step 3: Load model and run detection ---
model = YOLO(best_model_path)

# Lower conf threshold to detect subtle predictions (e.g., 0.05)
results = model(image_path, conf=0.05)

# --- Step 4: Extract and print detections ---
result = results[0]
names = model.names

if len(result.boxes) == 0:
    print("⚠️ No objects detected.")
else:
    print("✅ Detected Objects:")
    for i, (cls, conf) in enumerate(zip(result.boxes.cls, result.boxes.conf)):
        class_name = names[int(cls)]
        print(f"{i+1}. {class_name} ({conf:.2f})")

# --- Step 5: Draw bounding boxes and labels ---
annotated_image = result.plot()  # YOLO built-in visualization

# --- Step 6: Save annotated image ---
output_path = os.path.join(
    os.path.dirname(image_path),
    "sample_image_annotated.jpg"
)
cv2.imwrite(output_path, annotated_image)
print(f"💾 Annotated image saved at: {output_path}")

# --- Step 7: Display image ---
plt.figure(figsize=(7,7))
plt.imshow(cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("Detected Objects in Sample Image")
plt.show()
```
