```python
import os
import fitz
import tkinter as tk
from tkinter import filedialog
import google.generativeai as genai

# Configure Gemini API key
genai.configure(api_key="AIzaSyCQ_sJhXqEe7-SAadMPN981hbE_upB_MXw")

class PDFHelper:
    def __init__(self):
        self.model_id = "gemini-2.5-flash"
        self.session = None

    @staticmethod
    def select_pdf():
        """Open a file dialog for the user to choose a PDF file."""
        root = tk.Tk()
        root.withdraw()
        file_path = filedialog.askopenfilename(
            title="Select a PDF file",
            filetypes=[("PDF Files", "*.pdf")]
        )
        return file_path

    @staticmethod
    def get_pdf_text(file_path):
        """Extracts text content from the chosen PDF."""
        if not os.path.exists(file_path):
            print(f"File not found: {file_path}")
            return ""

        try:
            with fitz.open(file_path) as document:
                text_content = "".join(page.get_text("text") for page in document)

            if not text_content.strip():
                print("No readable text found. The PDF may be image-only.")
            else:
                print(f"Successfully read '{os.path.basename(file_path)}'.\n")

            return text_content.strip()
        except Exception as error:
            print(f"Error while reading PDF: {error}")
            return ""

    def start_session(self):
        """Initialize a Gemini chat session."""
        model = genai.GenerativeModel(self.model_id)
        self.session = model.start_chat(history=[])

    def chat_about_pdf(self, pdf_text):
        """Allow the user to ask questions about the extracted PDF text."""
        if not self.session:
            self.start_session()

        print("\nYou may now ask questions about the document.")
        print("Type 'exit' or 'quit' to end the session.\n")

        while True:
            user_input = input("Ask: ").strip()
            if user_input.lower() in {"exit", "quit"}:
                print("Session ended.")
                break

            query_prompt = (
                f"You are an intelligent assistant. Based on the following document, "
                f"answer the user's question accurately.\n\n"
                f"--- DOCUMENT START ---\n{pdf_text[:20000]}\n--- DOCUMENT END ---\n\n"
                f"Question: {user_input}"
            )

            try:
                response = self.session.send_message(query_prompt)
                print("\nAnswer:", response.text, "\n")
            except Exception as error:
                print(f"Error communicating with Gemini API: {error}")

def run_program():
    print("Please select a PDF file to analyze.\n")
    pdf_file = PDFHelper.select_pdf()

    if not pdf_file:
        print("No file selected. Exiting.")
        return

    print(f"Reading PDF file: {pdf_file}\n")
    content = PDFHelper.get_pdf_text(pdf_file)

    if content:
        print("Extracted text preview (first 500 characters):\n")
        print(content[:500])
        assistant = PDFHelper()
        assistant.chat_about_pdf(content)
    else:
        print("No readable text found in the selected PDF.")

if __name__ == "__main__":
    run_program()
```
